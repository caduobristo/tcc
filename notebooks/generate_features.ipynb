# Gerador de Features para Transformers

Este notebook pré-processa os áudios `.wav` originais em espectrogramas Log-Mel (usando a implementação *Kaldi Fbank*) otimizados para as taxas de amostragem específicas exigidas por cada arquitetura Transformer.

```python
import os
import sys

# Adiciona o diretório raiz ao sys.path para conseguirmos importar a pasta src
sys.path.append('..')
from src.data.feature_extraction import process_dataset_parallel
```

## Configuração do Processamento
Você pode rodar esse bloco múltiplas vezes variando o dataset e a taxa de amostragem (`TARGET_SR`).

```python
# --- SCENARIO SELECTOR ---
# Opções: 'mono_disc' | 'mono_cont' | 'poly_disc' | 'poly_cont'
DATASET_TYPE = 'poly_disc'

# --- TARGET CONFIGURATION ---
# 16000: Usado por AST e AudioMAE 
# 32000: Usado por PaSST e HTS-AT
TARGET_SR = 16000

if TARGET_SR == 16000:
    OUTPUT_FOLDER = 'mel_16'
elif TARGET_SR == 32000:
    OUTPUT_FOLDER = 'mel_32'
else:
    raise ValueError("Taxa de amostragem não mapeada!")

# Mapeamento de pastas do dataset
if DATASET_TYPE == 'mono_disc':
    DATASET_ROOT = '../dataset/GUITAR-FX-DIST/Mono_Discrete'
elif DATASET_TYPE == 'mono_cont':
    DATASET_ROOT = '../dataset/GUITAR-FX-DIST/Mono_Continuous'
elif DATASET_TYPE == 'poly_disc':
    DATASET_ROOT = '../dataset/GUITAR-FX-DIST/Poly_Discrete'
elif DATASET_TYPE == 'poly_cont':
    DATASET_ROOT = '../dataset/GUITAR-FX-DIST/Poly_Continuous'
else:
    raise ValueError(f"Tipo de dataset desconhecido: {DATASET_TYPE}")

print(f"Configurado para processar o dataset '{DATASET_TYPE}' a {TARGET_SR}Hz.")
print(f"Os novos arquivos .npy serão salvos na sub-pasta: {OUTPUT_FOLDER}/")
```

## Execução Paralela

```python
print("Iniciando conversão de áudios...")

process_dataset_parallel(
    dataset_root=DATASET_ROOT,
    target_sr=TARGET_SR,
    output_folder_name=OUTPUT_FOLDER,
    num_workers=8  # Usando 8 threads. Ajuste se desejar liberar mais espaço para uso do PC.
)

print("Processamento concluído com sucesso!")
```
