# AudioMAE - AudioSet Evaluation Template

This notebook is ready for full paper-style evaluation once AudioSet is prepared locally. It expects AST-style JSON files and the AudioSet label CSV under `experiments/audiomae/data/audioset`.

```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd().resolve()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
if not (ROOT / "repo").is_dir():
    raise RuntimeError("Launch this notebook from experiments/audiomae/notebooks or experiments/audiomae")
REPO = ROOT / "repo"
ENV_PYTHON = ROOT / ".conda" / "envs" / "audiomae" / "python.exe"
DATA = ROOT / "data" / "audioset"

paths = {
    "train_json": DATA / "train.json",
    "eval_json": DATA / "eval_19k.json",
    "label_csv": DATA / "class_labels_indices.csv",
    "finetuned_ckpt": ROOT / "ckpt" / "finetuned.pth",
}
for name, path in paths.items():
    print(f"{name}: {path} exists={path.exists()}")
```

```python
missing = [name for name, path in paths.items() if not path.exists()]
if missing:
    print("Missing files, skipping evaluation:", missing)
else:
    cmd = [
        str(ENV_PYTHON),
        str(REPO / "main_finetune_as.py"),
        "--model", "vit_base_patch16",
        "--dataset", "audioset",
        "--data_train", str(paths["train_json"]),
        "--data_eval", str(paths["eval_json"]),
        "--label_csv", str(paths["label_csv"]),
        "--finetune", str(paths["finetuned_ckpt"]),
        "--batch_size", "16",
        "--eval",
    ]
    print("Running:", " ".join(cmd))
    subprocess.run(cmd, cwd=str(REPO), check=True)
```
